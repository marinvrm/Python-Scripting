## Exercice 1 - Fusionner deux listes

Ecrire une fonction qui fusionne deux listes en une seule liste triee.

```python
def merge_lists(list1, list2):
    """Fusionne deux listes en une seule liste triee (ordre croissant)."""
    merged = list(list1) + list(list2)
    merged.sort()
    return merged


# Tests
a = [5, 2, 9, 1]
b = [7, 3, 8]

print("Liste 1 :", a)
print("Liste 2 :", b)
print("Fusion  :", merge_lists(a, b))
print("Liste 1 inchangee :", a)   # les listes d'origine ne sont pas modifiees
```

### Variante : fusion manuelle de deux listes deja triees

Si les deux listes sont **deja triees**, on peut les fusionner en une seule passe
(algorithme de fusion, comme dans le tri fusion) sans avoir a re-trier.

```python
def merge_sorted_lists(list1, list2):
    """Fusionne deux listes DEJA triees, en une seule passe."""
    result = []
    i = j = 0

    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1

    # On ajoute le reste (une seule des deux lignes ajoute vraiment quelque chose)
    result.extend(list1[i:])
    result.extend(list2[j:])
    return result


print(merge_sorted_lists([1, 4, 6], [2, 3, 10]))
print(merge_sorted_lists([], [2, 3]))
print(merge_sorted_lists([1, 2, 3], []))
```

## Exercice 2 - Supprimer les doublons

Supprimer les doublons d'une liste.

```python
def remove_duplicates(items):
    """Supprime les doublons en conservant l'ordre d'apparition."""
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result


nombres = [1, 3, 3, 5, 1, 7, 5, 9, 1]
mots = ["chat", "chien", "chat", "oiseau", "chien"]

print(remove_duplicates(nombres))
print(remove_duplicates(mots))
```

### Variantes

- `set(...)` : le plus court, mais **l'ordre n'est pas conserve**.
- `dict.fromkeys(...)` : conserve l'ordre (les dictionnaires sont ordonnes depuis Python 3.7).

```python
nombres = [1, 3, 3, 5, 1, 7, 5, 9, 1]

# 1) Avec un set : ordre non garanti
print("set        :", list(set(nombres)))

# 2) Avec dict.fromkeys : ordre conserve, en une ligne
print("fromkeys   :", list(dict.fromkeys(nombres)))

# 3) Version triee
print("set + tri  :", sorted(set(nombres)))
```

## Exercice 3 - Liste strictement croissante

Ecrire un programme qui verifie si une liste est strictement croissante
(chaque element est strictement superieur au precedent).

```python
def is_strictly_increasing(items):
    """Retourne True si chaque element est strictement superieur au precedent."""
    for i in range(1, len(items)):
        if items[i] <= items[i - 1]:
            return False
    return True


tests = [
    [1, 2, 3, 4, 5],   # croissante
    [1, 2, 2, 4],      # egalite -> pas STRICTEMENT croissante
    [5, 4, 3],         # decroissante
    [42],              # un seul element -> True
    [],                # liste vide -> True (aucune paire ne viole la regle)
]

for t in tests:
    print(f"{str(t):<18} -> {is_strictly_increasing(t)}")
```

### Variante en une ligne

`all(...)` combine a `zip(items, items[1:])` compare chaque element au suivant.

```python
def is_strictly_increasing_v2(items):
    return all(a < b for a, b in zip(items, items[1:]))


for t in tests:
    print(f"{str(t):<18} -> {is_strictly_increasing_v2(t)}")

# Les deux versions doivent donner exactement les memes resultats
print()
print("Memes resultats :",
      all(is_strictly_increasing(t) == is_strictly_increasing_v2(t) for t in tests))
```

## Exercice 4 - Filtrer une liste de mots

Parcourir une liste de mots et creer une nouvelle liste contenant uniquement les mots qui :

- commencent par une voyelle (a, e, i, o, u),
- **et** ont une longueur superieure a 5 caracteres.

```python
VOWELS = set("aeiouAEIOU")

words = [
    "apple", "elephant", "banana", "orange", "kiwi",
    "umbrella", "igloo", "avocado", "sun", "engine",
    "Ostrich", "idea", "unique",
]


def filter_words(words):
    """Garde les mots commencant par une voyelle ET de longueur > 5."""
    return [w for w in words if len(w) > 5 and w[0] in VOWELS]


print("Liste de depart :", words)
print("Resultat        :", filter_words(words))
```

### Pourquoi cette ligne est efficace

- `len(w) > 5` est teste **en premier** : c'est le test le moins couteux, et grace au
  court-circuit du `and` il evite d'aller lire `w[0]` pour les mots trop courts
  (ce qui protege aussi des chaines vides, ou `w[0]` leverait une `IndexError`).
- `VOWELS` est un `set` contenant deja les majuscules : le test d'appartenance est
  en O(1) et on economise un appel a `.lower()` par mot.

```python
print(f"{'mot':<12} {'len > 5':<9} {'voyelle':<9} garde")
print("-" * 42)
for word in words:
    long_enough = len(word) > 5
    starts_vowel = word[0] in VOWELS
    print(f"{word:<12} {str(long_enough):<9} {str(starts_vowel):<9} {long_enough and starts_vowel}")
```

### Variante : la boucle explicite

Exactement le meme resultat, ecrit en plusieurs lignes.

```python
def filter_words_v2(words):
    result = []
    for word in words:
        if len(word) > 5 and word[0] in VOWELS:
            result.append(word)
    return result


print(filter_words_v2(words))
print("Meme resultat :", filter_words(words) == filter_words_v2(words))

# Et si on voulait les deux criteres separement :
print("Commencent par une voyelle :", [w for w in words if w[0] in VOWELS])
print("Plus de 5 caracteres       :", [w for w in words if len(w) > 5])
```
